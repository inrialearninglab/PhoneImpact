```python
import pandas as pd

Fer = [10155, 3423, 1090]
Cuivre = [17.32, 3922, 11451]
Or = [0.00236, 4051, 13870]

df = pd.DataFrame(
    [Or, Cuivre, Fer],
    columns=["Stériles miniers", "Résidus miniers", "Métal produit"],
    index=["Or", "Cuivre", "Fer"]
)

df = df.transpose()
```

```python
import matplotlib.pyplot as plt
from pywaffle import Waffle
```

```python
colors = ['#deb305', '#8d4723', '#b1b1b1']
label = [df.index[2-i] + f": {int(df.Fer[2-i])} tonnes" for i in range(len(df.Fer))]
```

```python
fig = plt.figure(FigureClass=Waffle, 
                 rows=30, columns=30,
                 values=df.Cuivre, 
                 colors=colors,
                 labels=label,
                 figsize = (15,6),
                 legend={'loc':'upper right',
                         'bbox_to_anchor': (1.5, 1),
                        },
                 vertical=True
                )
plt.tight_layout()
plt.title("Extraction du cuivre", size=20, pad=20)
plt.show()
```

```python
fig = plt.figure(FigureClass=Waffle, 
                 rows=30, 
                 columns=30,
                 values=df.Fer, 
                 colors=colors,
                 labels=label,
                 figsize = (15,6),
                 legend={'loc':'upper right',
                         'bbox_to_anchor': (1.32, 1),
                        },
                 block_arranging_style='new-line',
                 vertical=True
                 )
plt.tight_layout()
plt.title("Extraction du fer", size=20)
plt.show()
```

```python
from pywaffle import Waffle

fig = plt.figure(FigureClass=Waffle, 
                 rows=30, columns=30,
                 values=df.Or, 
                 colors=colors,
                 labels=label,
                 figsize = (15,6),
                 legend={'loc':'upper right',
                         'bbox_to_anchor': (1.32, 1),
                        })
plt.tight_layout()
plt.title("Extraction de l'or", size=20)
plt.show()
```
